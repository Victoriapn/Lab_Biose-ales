```python

import numpy as np
import matplotlib.pyplot as plt
import neurokit2 as nk
from ecgdetectors import Detectors

# ecg: vector con la señal registrada
# fs: frecuencia de muestreo real en Hz
ecg = np.asarray(ecg, dtype=float).ravel()

# NeuroKit2
ecg_limpio = nk.ecg_clean(ecg, sampling_rate=fs)
_, info = nk.ecg_peaks(
    ecg_limpio,
    sampling_rate=fs,
    correct_artifacts=False
)
picos_nk = np.asarray(info["ECG_R_Peaks"])

# Detectors: algoritmo Pan-Tompkins
detector = Detectors(fs)
picos_pt = np.asarray(
    detector.pan_tompkins_detector(ecg)
)

# Representación
tiempo = np.arange(len(ecg)) / fs

plt.figure(figsize=(12, 4))
plt.plot(tiempo, ecg, label="ECG", alpha=0.7)
plt.scatter(tiempo[picos_nk], ecg[picos_nk],
            label="NeuroKit2", marker="o")
plt.scatter(tiempo[picos_pt], ecg[picos_pt],
            label="Pan-Tompkins", marker="x")
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud")
plt.title("Comparación de detectores de picos R")
plt.legend()
plt.grid()
plt.show()

print("Picos NeuroKit2:", len(picos_nk))
print("Picos Pan-Tompkins:", len(picos_pt))
```
