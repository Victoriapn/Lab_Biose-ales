1) Escribir una función que reciba un diccionario con las notas de los alumnos de un curso y devuelva
una serie con la nota mínima, la máxima, media y la desviación típica.


```python
def estadisticas_notas(diccionario_notas):
    """
    Recibe un diccionario con notas de alumnos y devuelve una Serie de Pandas
    con la nota mínima, máxima, media y desviación típica.
    """
    notas = pd.Series(diccionario_notas)

    # Calculamos las estadísticas y las guardamos en una nueva Serie
    estadisticas = pd.Series({
        'Mínima': notas.min(),
        'Máxima': notas.max(),
        'Media': notas.mean(),
        'Desviación típica': notas.std()
    })

    return estadisticas
```

```python
notas_alumnos = {'Juan': 8.0, 'María': 9.5, 'Pedro': 6.0, 'Ana': 7.5}
print(estadisticas_notas(notas_alumnos))
```

2) Cálculo del Índice de Masa Corporal (IMC) con Pandas.

    a) Cargar los datos desde el archivo "datos.csv".

    b) Mostrar las primeras y últimas filas del DataFrame para conocer la estructura de los datos.

    c) Borrar la columna "Unnamed: 0" del DataFrame.

```python
df = pd.read_csv("datos (1).csv", sep=';')

print(" Primeras 5 filas ")
print(df.head())

print("\n Últimas 5 filas ")
print(df.tail())

# c) Borrar la columna "Unnamed: 0" del DataFrame
df = df.drop(columns=['Unnamed: 0'])
```

    d y e) Calcular el IMC y agregarlo como una nueva columna llamada "BMI"

```python
# Convertimos la altura de centímetros a metros
df['Altura_m'] = df['Height'] / 100

# Aplicamos la fórmula del IMC: Peso (kg) / (Altura (m))^2
df['BMI'] = df['Weight'] / (df['Altura_m'] ** 2)

print("\n DataFrame resultante con el IMC ")
print(df.head())
```

f) Categorizar a los participantes según su IMC en:

• Bajo peso (< 18.5)

• Normal (18.5 - 24.9)

• Sobrepeso (25 - 29.9)

• Obesidad (≥ 30)

```python
# Definimos los límites y nombres
limites = [0, 18.5, 25.0, 30.0, float('inf')]
etiquetas = ['Bajo peso', 'Normal', 'Sobrepeso', 'Obesidad']

df['Categoria_IMC'] = pd.cut(df['BMI'], bins=limites, labels=etiquetas, right=False)

# Mostrar el resultado final con las columnas más importantes
print("\n DataFrame resultante con BMI y Categoría ")
print(df[['Gender', 'Weight', 'Height', 'BMI', 'Categoria_IMC']].head(10))
```
